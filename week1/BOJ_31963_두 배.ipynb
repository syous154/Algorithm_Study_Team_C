# 📝 문제 정보 (BOJ 31963)
* 문제 이름: 두 배
* 문제 링크: https://www.acmicpc.net/problem/31963
* 난이도: Gold 4
* 알고리즘 유형: 그리디 알고리즘, 수학

## 문제 설명
길이 $N$인 양의 정수열 $A_1, \dots , A_N$이 주어진다. 이 수열을 오름차순으로 만들려 한다. 수열 $A_1, \dots , A_N$이 오름차순이라는 것은, 각 $i$ ($1 ≤ i ≤ N - 1$)에 대해 $A_i ≤ A_{i+1}$이라는 것이다.
수열 $A$를 오름차순으로 만들기 위해, 수열 $A$에 다음 연산을 몇 번이든 반복해서 적용할 수 있다.

어떤 $i$ ($1 ≤ i ≤ N$)에 대해 $A_i$에 $2$를 곱한다.

연산을 최소 횟수로 적용해서 $A$를 오름차순으로 만들고 싶다. 이때, 최소 횟수를 구하라.

### 입력
첫 번째 줄에 $N$이 주어진다.
두 번째 줄에 $A_1, \dots , A_N$이 주어진다.

### 출력
첫 번째 줄에 답을 출력한다.

### 예제 입력 1
```
입력:
5
3 1 4 1 5
출력:
4
```

### 예제 입출력 2
```
입력:
5
3 1 5 1 5
출력:
6
```

### 예제 입출력 3
```
입력:
5
1 2 3 4 5
출력:
0
```


# 💡 접근 방법
1. $A_{i-1}$ $\leq$ $A_i \times 2^x$ 를 만족하는 $x$의 최소값은 $log_2$ $\frac{A_{i-1}}{A_i}$
2. x가 최소 연산 횟수이므로 이를 `cntlist`에 저장하여 중복 연산을 피하면 100점까지 얻을 수 있다.

# ⏱️ 시간복잡도
$O(N)$   
$\because$ for문 반복

# 🗃️ 공간복잡도
$O(N)$   
$\because$ `cntlist` 리스트 사용


# ✏️ 풀이

````python
import sys
import math
input = sys.stdin.readline

n = int(input())
list1 = list(map(int, input().split()))

cnt = 0
cntlist = [0]*n
for i in range(1, n):
    ratio = math.ceil(math.log2(list1[i - 1] / list1[i])) + cntlist[i - 1]
    if ratio > 0:
        cntlist[i] = max(0, ratio)
        cnt += cntlist[i]
    
print(cnt)
````

# 🧐 배운 점 & 어려웠던 점
- 리스트 값을 직접 변경하지 말고 별도의 리스트에 연산 횟수만 저장하도록 하자.
