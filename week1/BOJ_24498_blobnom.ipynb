# 📝 문제 정보 (BOJ 24498)
* 문제 이름: blobnom
* 문제 링크: https://www.acmicpc.net/problem/24498
* 난이도: Silver 3
* 알고리즘 유형: 그리디 알고리즘

## 문제 설명
블롭들은 심심해서 서로를 이용해 $N$개의 탑을 만들었다. 각 탑의 높이는 그 탑에 있는 블롭의 수와 같다.

여러분은 다음 행동을 $0$회 이상 할 수 있다.

처음과 마지막이 아닌 탑 중 하나를 선택한다. 단, 선택한 탑과 인접한 두 탑의 높이가 모두 $1$ 이상이어야 한다.
선택한 탑과 인접한 두 탑에 있는 블롭을 한 마리씩 각각 땅에 내려놓는다. 즉, 인접한 두 탑의 높이가 모두 $1$만큼 감소한다.
땅에 내려놓은 두 마리의 블롭 중 하나의 블롭만 1.에서 선택한 탑에 쌓는다. 즉, 선택한 탑의 높이가 $1$만큼 증가한다.

이 과정에서 이전에 인접하지 않았던 두 탑이 새롭게 인접하게 되지는 않는다. 채완이를 위해 만들 수 있는 가장 높은 탑의 높이를 구해 주자.

### 입력
첫째 줄에 탑의 개수 $N$이 주어진다.
둘째 줄에 각 탑의 높이 $A_1, A_2, \cdots, A_N$이 공백으로 구분되어 주어진다.

### 출력
문제의 정답을 출력한다.

### 예제 입력 1
```
입력:
4
1 3 2 2
출력:
4
```

### 예제 입출력 2
```
입력:
5
2 4 3 5 5
출력:
8
```


# 💡 접근 방법
1. 리스트에서 가장 큰 값을 답으로 초기화
2. i = 1 ~ n-2를 탐색하면서 양 옆의 높이가 1 이상인 곳을 탐색하고, 그곳에서 (현재 탑 높이 + 양 옆 중 작은 탑의 높이)로 답을 업데이트

# ⏱️ 시간복잡도
$O(N)$   
$\because$ for문 반복 & `max(blob)` 연산

# 🗃️ 공간복잡도
$O(N)$   
$\because$ `blob` 리스트 사용

# ✏️ 풀이

````python
import sys
input = sys.stdin.readline

n = int(input())
blob = list(map(int, input().split()))

max1 = max(blob)
for i in range(1, n-1):
    if blob[i-1] >= 1 and blob[i+1] >= 1:
        max1 = max(max1, blob[i]+min(blob[i-1], blob[i+1]))

print(max1)
````
