## 문제 설명
지원이에게 2진 수열을 가르쳐 주기 위해, 지원이 아버지는 그에게 타일들을 선물해주셨다. 그리고 이 각각의 타일들은 0 또는 1이 쓰여 있는 낱장의 타일들이다.
어느 날 짓궂은 동주가 지원이의 공부를 방해하기 위해 0이 쓰여진 낱장의 타일들을 붙여서 한 쌍으로 이루어진 00 타일들을 만들었다. 결국 현재 1 하나만으로 이루어진 타일 또는 0타일을 두 개 붙인 한 쌍의 00타일들만이 남게 되었다.
그러므로 지원이는 타일로 더 이상 크기가 N인 모든 2진 수열을 만들 수 없게 되었다. 예를 들어, N=1일 때 1만 만들 수 있고, N=2일 때는 00, 11을 만들 수 있다. (01, 10은 만들 수 없게 되었다.) 또한 N=4일 때는 0011, 0000, 1001, 1100, 1111 등 총 5개의 2진 수열을 만들 수 있다.
우리의 목표는 N이 주어졌을 때 지원이가 만들 수 있는 모든 가짓수를 세는 것이다. 단 타일들은 무한히 많은 것으로 가정하자.

### 입력
첫 번째 줄에 자연수 N이 주어진다. (1 ≤ N ≤ 1,000,000)

### 출력
첫 번째 줄에 지원이가 만들 수 있는 길이가 N인 모든 2진 수열의 개수를 15746으로 나눈 나머지를 출력한다.

### 예제 입력 1
```
입력:
4
출력:
5
```


# 💡 접근 방법
#### 1. dp 점화식 찾기
N 길이의 타일을 만드려면 두 가지 방법이 있다.
1) N-1 길이의 타일에 1 추가하기
2) N-2 길이의 타일에 00 추가하기
따라서 결국 N 길이의 타일을 만드는 방법의 수는 N-1 길이의 타일을 만드는 방법의 수 + N-2 길이의 타일을 만드는 방법의 수와 같다.      
   
→ `dp[n] = dp[n-1] + dp[n-2]`

#### 2. 모듈러 연산 이용하기
위의 점화식을 그대로 사용하면 n이 커지면서 dp 배열에 저장되는 수의 크기가 지나치게 커지게 된다.   
그러므로 `dp[n] = (dp[n-1] + dp[n-2])%15764` 로 저장하여 메모리 초과를 방지한다.

# ⏱️ 시간복잡도
$O(N)$   
$\because$ dp 배열 사용

# 🗃️ 공간복잡도
$O(N)$   
$\because$ n번 반복되는 for문 사용


# ✏️ 풀이

````python
n = int(input())

if n == 1:
    print(1)
    exit()
elif n == 2:
    print(2)
    exit()
else:  # n >= 3
    dp = [0]*(n+1)
    dp[1] = 1
    dp[2] = 2

    for i in range(3, n+1):
        dp[i] = (dp[i-2] + dp[i-1])%15746

    print(dp[n])
````

# 🧐 배운 점 & 어려웠던 점
- 점화식 찾는 훈련을 많이 하자.
- 메모리를 효율적으로 사용하는 방법을 항상 고민하자. 모듈러 연산을 기억하자.
- 예외 처리에 유의하자. `n = 1`인 상황에서는 `dp[2] = 2`가 동작할 수 없다. (`IndexError`)
