# 📝 문제 정보 (BOJ 1238)
* 문제 이름: 파티
* 문제 링크: https://www.acmicpc.net/problem/1238
* 난이도: Gold 3
* 알고리즘 유형: 데이크스트라, 그래프 이론, 최단 경로

## 문제 설명
N개의 숫자로 구분된 각각의 마을에 한 명의 학생이 살고 있다.
어느 날 이 N명의 학생이 X (1 ≤ X ≤ N)번 마을에 모여서 파티를 벌이기로 했다. 이 마을 사이에는 총 M개의 단방향 도로들이 있고 i번째 길을 지나는데 Ti(1 ≤ Ti ≤ 100)의 시간을 소비한다.
각각의 학생들은 파티에 참석하기 위해 걸어가서 다시 그들의 마을로 돌아와야 한다. 하지만 이 학생들은 워낙 게을러서 최단 시간에 오고 가기를 원한다.
이 도로들은 단방향이기 때문에 아마 그들이 오고 가는 길이 다를지도 모른다. N명의 학생들 중 오고 가는데 가장 많은 시간을 소비하는 학생은 누구일지 구하여라.

### 입력
첫째 줄에 N(1 ≤ N ≤ 1,000), M(1 ≤ M ≤ 10,000), X가 공백으로 구분되어 입력된다. 두 번째 줄부터 M+1번째 줄까지 i번째 도로의 시작점, 끝점, 그리고 이 도로를 지나는데 필요한 소요시간 Ti가 들어온다. 시작점과 끝점이 같은 도로는 없으며, 시작점과 한 도시 A에서 다른 도시 B로 가는 도로의 개수는 최대 1개이다.
모든 학생들은 집에서 X에 갈수 있고, X에서 집으로 돌아올 수 있는 데이터만 입력으로 주어진다.

### 출력
첫 번째 줄에 N명의 학생들 중 오고 가는데 가장 오래 걸리는 학생의 소요시간을 출력한다.

### 예제 입력 1
```
입력:
4 8 2
1 2 4
1 3 2
1 4 7
2 1 1
2 3 5
3 1 2
3 4 4
4 2 3
출력:
10
```


# 💡 접근 방법
1. 단방향 도로, 간선마다 다른 가중치 → 다익스트라 알고리즘
2. 집 → 파티, 파티 → 집으로 2번 다익스트라를 돌려야 한다.

# ⏱️ 시간복잡도
$O(N \times (N+M) \times log \,N)$   
다익스트라 알고리즘의 시간 복잡도는 $O((정점의 \,개수 + 간선의\, 개수) \times log\, 정점의 \,개수)$   
이를 N+1번 반복

# 🗃️ 공간복잡도
$O(N+M)$   
`graph` 사용

# ✏️ 풀이

````python
import sys
import heapq
input = sys.stdin.readline

n, m, x = map(int, input().split())
graph = [[] for _ in range(n+1)]
for _ in range(m):
    a, b, c = map(int, input().split())
    graph[a].append([b, c])

def dijkstra(graph, start):
    times = [float('inf')]*(n+1)     
    times[start] = 0
    q = []
    heapq.heappush(q, [times[start], start])
    while q:
        t, node = heapq.heappop(q)
        if times[node] < t:
            continue
        for next_node, next_time in graph[node]:
            time = t + next_time
            if time < times[next_node]:
                times[next_node] = time
                heapq.heappush(q, [time, next_node])
    return times
    
# come back home
home = dijkstra(graph, x)

# to the party
party = [0]    # 0th index is dummy data
for i in range(1, n+1):
    p = dijkstra(graph, i)
    party.append(p[x])
    
answer = [x+y for x, y in zip(home, party) if x + y != float('inf')]
print(max(answer))
````

# 🧐 배운 점 & 어려웠던 점
- 파티로 가는 과정을 그래프를 뒤집어서 다익스트라 알고리즘을 쓰면 n번 반복하지 않고 1번에 처리할 수 있다.
