## 문제 설명
옛날 옛적에 수학이 항상 큰 골칫거리였던 나라가 있었다. 이 나라의 국왕 김지민은 다음과 같은 문제를 내고 큰 상금을 걸었다.
길이가 N인 정수 배열 A와 B가 있다. 다음과 같이 함수 S를 정의하자.
S = A[0] × B[0] + ... + A[N-1] × B[N-1]
S의 값을 가장 작게 만들기 위해 A의 수를 재배열하자. 단, B에 있는 수는 재배열하면 안 된다.
S의 최솟값을 출력하는 프로그램을 작성하시오.

### 입력
첫째 줄에 N이 주어진다. 둘째 줄에는 A에 있는 N개의 수가 순서대로 주어지고, 셋째 줄에는 B에 있는 수가 순서대로 주어진다. N은 50보다 작거나 같은 자연수이고, A와 B의 각 원소는 100보다 작거나 같은 음이 아닌 정수이다.

### 출력
첫째 줄에 S의 최솟값을 출력한다.

### 예제 입력 1
```
입력:
5
1 1 1 6 0
2 7 8 3 1
출력:
18
```

### 예제 입출력 2
```
입력:
3
1 1 3
10 30 20
출력:
80
```

### 예제 입출력 3
```
입력:
9
5 15 100 31 39 0 0 3 26
11 12 13 2 3 4 5 9 1
출력:
528
```


# 💡 접근 방법
1. 두 리스트 모두 정렬후 zip 합수를 통해 곱셈 구현
2. 
3. 

# ⏱️ 시간복잡도
O(nlog(n))

# 🗃️ 공간복잡도
O(n)


````python
# 코드 작성
def solution():
    _ = int(input())
    num_list1 = map(int, input().split()) # 리스트1
    num_list2 = map(int, input().split()) # 리스트2
    num_list1 = sorted(num_list1)
    num_list2 = sorted(num_list2, reverse=True)
    answer = 0
    # 최솟값을 구하기 위해 최대값과 최소값을 곱하는 방식으로 구현
    for num1, num2 in zip(num_list1, num_list2):
        answer += num1 * num2
    # 여기에 코드를 작성하세요
    print(answer)

# 입력 처리
if __name__ == "__main__":
    solution()
````

# 🧐 배운 점 & 어려웠던 점
- 이 방식이 그리디인가?
  - Yes -> 최적해를 구하는 과정에서 가능한 모든 경우의 수를 탐색하는 것이 아니라 낮은것과 높은것을 곱해서 더하는 방식으로 최적화를 진행했을 뿐 큰틀은 벗어나지 않음.
