# 📝 문제 정보 (BOJ 2839)

* 알고리즘 유형: 다이나믹 프로그래밍, 그리디 알고리즘, 수학

## 문제 설명
상근이는 요즘 설탕공장에서 설탕을 배달하고 있다. 상근이는 지금 사탕가게에 설탕을 정확하게 N킬로그램을 배달해야 한다. 설탕공장에서 만드는 설탕은 봉지에 담겨져 있다. 봉지는 3킬로그램 봉지와 5킬로그램 봉지가 있다.
상근이는 귀찮기 때문에, 최대한 적은 봉지를 들고 가려고 한다. 예를 들어, 18킬로그램 설탕을 배달해야 할 때, 3킬로그램 봉지 6개를 가져가도 되지만, 5킬로그램 3개와 3킬로그램 1개를 배달하면, 더 적은 개수의 봉지를 배달할 수 있다.
상근이가 설탕을 정확하게 N킬로그램 배달해야 할 때, 봉지 몇 개를 가져가면 되는지 그 수를 구하는 프로그램을 작성하시오.

### 입력
첫째 줄에 N이 주어진다. (3 ≤ N ≤ 5000)

### 출력
상근이가 배달하는 봉지의 최소 개수를 출력한다. 만약, 정확하게 N킬로그램을 만들 수 없다면 -1을 출력한다.

### 예제 입력 1
```
입력:
18
출력:
4
```

### 예제 입출력 2
```
입력:
4
출력:
-1
```

### 예제 입출력 3
```
입력:
6
출력:
2
```

### 예제 입출력 4
```
입력:
9
출력:
3
```

### 예제 입출력 5
```
입력:
11
출력:
3
```


# 💡 접근 방법
1. 5kg가 최대한 많이 가능한 방식으로 구현
2. 5kg로 가능한 범위까지 진행 후 3kg로 진행
3. 5와3의 덧셈으로 가능한 방법을 최소값을 기준으로 탐색.

# ⏱️ 시간복잡도
O(n)

# 🗃️ 공간복잡도
O(1)


````python
# 코드 작성
def solution():
    # 여기에 코드를 작성하세요
    n = int(input()) # 입력
    temp = n//5 # 5로 나눈 몫
    answer = -1
    while temp >= 0: # 0이 될때까지 반복
        if (n - temp*5)%3 == 0: # 5로 나눈 나머지를 3으로 나눈 나머지가 0이면
            answer = temp + (n - temp*5)//3 # 5로 나눈 몫 + 5로 나눈 나머지를 3으로 나눈 몫
            break
        temp -= 1 # 5로 나눈 몫을 1씩 감소
    print(answer) #
""
# 입력 처리
if __name__ == "__main__":
    solution()
````
