# 📝 문제 정보 (BOJ 17484)
* 문제 이름: 진우의 달 여행 (Small)
* 문제 링크: https://www.acmicpc.net/problem/17484
* 난이도: Silver 3
* 알고리즘 유형: 브루트포스 알고리즘, 다이나믹 프로그래밍

## 문제 설명
우주비행이 꿈이였던 진우는 음식점 '매일매일싱싱'에서 열심히 일한 결과 달 여행에 필요한 자금을 모두 마련하였다! 지구와 우주사이는 N X M 행렬로 나타낼 수 있으며 각 원소의 값은 우주선이 그 공간을 지날 때 소모되는 연료의 양이다.
[예시]

진우는 여행경비를 아끼기 위해 조금 특이한 우주선을 선택하였다. 진우가 선택한 우주선의 특징은 아래와 같다.
1. 지구 -> 달로 가는 경우 우주선이 움직일 수 있는 방향은 아래와 같다.

2. 우주선은 전에 움직인 방향으로 움직일 수 없다. 즉, 같은 방향으로 두번 연속으로 움직일 수 없다.
진우의 목표는 연료를 최대한 아끼며 지구의 어느위치에서든 출발하여 달의 어느위치든 착륙하는 것이다.
최대한 돈을 아끼고 살아서 달에 도착하고 싶은 진우를 위해 달에 도달하기 위해 필요한 연료의 최소값을 계산해 주자.

### 입력
첫줄에 지구와 달 사이 공간을 나타내는 행렬의 크기를 나타내는 N, M (2≤ N, M ≤ 6)이 주어진다.
다음 N줄 동안 각 행렬의 원소 값이 주어진다. 각 행렬의 원소값은 100 이하의 자연수이다.

### 출력
달 여행에 필요한 최소 연료의 값을 출력한다.

### 예제 입력 1
```
입력:
6 4
5 8 5 1
3 5 8 4
9 77 65 5
2 1 5 2
5 98 1 5
4 95 67 58
출력:
29
```

# ⏱️ 시간복잡도
O(nxm)

# 🗃️ 공간복잡도
O(nxm)


````python
def solution():
    # 입력 받기
    n, m = map(int, input().split())
    grid = []
    for _ in range(n):
        grid.append(list(map(int, input().split())))
    
    # dp[i][j][d] = i행 j열에 d 방향으로 도착했을 때의 최소 연료 소비량
    # d = 0: 왼쪽 아래 대각선, d = 1: 바로 아래, d = 2: 오른쪽 아래 대각선
    dp = [[[float('inf') for _ in range(3)] for _ in range(m)] for _ in range(n)]
    
    # 출발점 초기화
    for j in range(m):
        for d in range(3):
            # 왼쪽 끝에서는 왼쪽 대각선으로 이동 불가
            if j == 0 and d == 0:
                dp[0][j][d] = float('inf')
            # 오른쪽 끝에서는 오른쪽 대각선으로 이동 불가
            elif j == m-1 and d == 2:
                dp[0][j][d] = float('inf')
            else:
                dp[0][j][d] = grid[0][j]

    for i in range(1, n):
        for j in range(m):
            # 왼쪽 아래 대각선으로 이동 (d = 0)
            if j > 0:  # 좌측 경계 체크
                dp[i][j][0] = min(dp[i-1][j-1][1], dp[i-1][j-1][2]) + grid[i][j]
            
            # 바로 아래로 이동 (d = 1)
            dp[i][j][1] = min(dp[i-1][j][0], dp[i-1][j][2]) + grid[i][j]
            
            # 오른쪽 아래 대각선으로 이동 (d = 2)
            if j < m-1:  # 우측 경계 체크
                dp[i][j][2] = min(dp[i-1][j+1][0], dp[i-1][j+1][1]) + grid[i][j]
    
    # 최솟값 탐색
    min_fuel = float('inf')
    for j in range(m):
        for d in range(3):
            min_fuel = min(min_fuel, dp[n-1][j][d])
    
    print(min_fuel)

# 입력 처리
if __name__ == "__main__":
    solution()
````
