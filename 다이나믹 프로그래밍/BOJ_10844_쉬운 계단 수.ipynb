* 문제 링크: https://www.acmicpc.net/problem/10844

* 알고리즘 유형: 다이나믹 프로그래밍

## 문제 설명
45656이란 수를 보자.
이 수는 인접한 모든 자리의 차이가 1이다. 이런 수를 계단 수라고 한다.
N이 주어질 때, 길이가 N인 계단 수가 총 몇 개 있는지 구해보자. 0으로 시작하는 수는 계단수가 아니다.

### 입력
첫째 줄에 N이 주어진다. N은 1보다 크거나 같고, 100보다 작거나 같은 자연수이다.

### 출력
첫째 줄에 정답을 1,000,000,000으로 나눈 나머지를 출력한다.

### 예제 입력 1
```
입력:
1
출력:
9
```

### 예제 입출력 2
```
입력:
2
출력:
17
```

# ⏱️ 시간복잡도
O(n)

# 🗃️ 공간복잡도
O(n)


````python
# 코드 작성
def solution():
    # 여기에 코드를 작성하세요
    n = int(input())
    dp = [[0]* 10 for _ in range(n+1)] # n자리, 마지막이 j인 수의 개수표현 not 숫자 
    
    # 초기값 셋팅
    for i in range(1, 10): # 처음에 0은 제외 1~9만 
        dp[1][i] = 1    
    
    # DP 
    for i in range(2, n+1): # n번쨰까지 진행
        for j in range(10): # 0~9까지
            if j == 0: # 0일때는 1만 가능 
                dp[i][j] = dp[i-1][1]
            elif j == 9: # 9일때는 8만 가능
                dp[i][j] = dp[i-1][8]
            else: # 1~8일때는 1작은수, 1큰수 가능
                dp[i][j] = dp[i-1][j-1] + dp[i-1][j+1] % 1_000_000_000
    
    print(sum(dp[n]) % 1_000_000_000)

# 입력 처리
if __name__ == "__main__":
    solution()
````
