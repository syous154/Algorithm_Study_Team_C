# ⏱️ 시간복잡도
O(N+E)

# 🗃️ 공간복잡도
O(N)


````python
from collections import deque
from collections import defaultdict
import sys
# 코드 작성
def solution():
    input = sys.stdin.readline
    # 여기에 코드를 작성하세요
    node = int(input())
    visited = [False] * (node+1)
    
    # Tree 구성
    tree = defaultdict(list)
    for _ in range(node-1):
        a, b = map(int, input().split())
        tree[a].append(b)
        tree[b].append(a)
    # 1은 root 노드이므로 1로 출발.
    visited[1] = True
    q = deque([1])
    # BFS
    while q:
        current = q.popleft()
        for i in tree[current]:
            if not visited[i]:
                visited[i] = current
                q.append(i)
    for i in range(2, node+1):
        print(visited[i])
    pass

# 입력 처리
if __name__ == "__main__":
    solution()
````
